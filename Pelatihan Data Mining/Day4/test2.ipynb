```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("latihan4.csv")
df = pd.DataFrame(data)
```

```python
# mengisi data kosong

df["Usia"] = df["Usia"].fillna(df["Usia"].mean())

df["Pendapatan_Bulanan"] = df["Pendapatan_Bulanan"].fillna(
    df["Pendapatan_Bulanan"].mean()
)

df["Status_Pekerjaan"] = df["Status_Pekerjaan"].fillna(df["Status_Pekerjaan"].mode()[0])

df["Lama_Bekerja"] = df["Lama_Bekerja"].fillna(df["Lama_Bekerja"].mode()[0])

df["Status_Pernikahan"] = df["Status_Pernikahan"].fillna(
    df["Status_Pernikahan"].mode()[0]
)

df["Jumlah_Tanggungan"] = df["Jumlah_Tanggungan"].fillna(
    df["Jumlah_Tanggungan"].mode()[0]
)

df["Skor_Kredit"] = df["Skor_Kredit"].fillna(df["Skor_Kredit"].mean())

df["Riwayat_Kredit"] = df["Riwayat_Kredit"].fillna(df["Riwayat_Kredit"].mode()[0])

df["Kelayakan_Kredit"] = df["Kelayakan_Kredit"].fillna(df["Kelayakan_Kredit"].mode()[0])
```

```python
df.isnull().sum()
```

```python
df.head()
```

```python
df["Status_Pekerjaan"] = df["Status_Pekerjaan"].str.title()

df["Status_Pernikahan"] = df["Status_Pernikahan"].str.title()

df["Riwayat_Kredit"] = df["Riwayat_Kredit"].str.title()

df.head()
```

```python
Q1 = df["Pendapatan_Bulanan"].quantile(0.25)
Q3 = df["Pendapatan_Bulanan"].quantile(0.75)
IQR = Q3 - Q1

upper_bound = Q3 + 1.5 * IQR
lower_bound = Q1 - 1.5 * IQR

outliers = df[
    (df["Pendapatan_Bulanan"] < lower_bound) | (df["Pendapatan_Bulanan"] > upper_bound)
]
```

```python
Q1 = df["Skor_Kredit"].quantile(0.25)
Q3 = df["Skor_Kredit"].quantile(0.75)
IQR = Q3 - Q1

upper_bound = Q3 + 1.5 * IQR
lower_bound = Q1 - 1.5 * IQR

outliers = df[(df["Skor_Kredit"] < lower_bound) | (df["Skor_Kredit"] > upper_bound)]
```

```python
df["Kelayakan_Kredit"] = df["Kelayakan_Kredit"].replace("layak", "Layak")
df["Kelayakan_Kredit"] = df["Kelayakan_Kredit"].replace("TIDAK", "Tidak")
```

```python
df["Kelayakan_Kredit"].unique()
```

```python
from sklearn.tree import DecisionTreeClassifier

df_encoded = df.copy()

df_encoded["Status_Pekerjaan"] = df_encoded["Status_Pekerjaan"].map(
    {"Wirausaha": 4, "Tetap": 3, "Wiraswasta": 2, "Kontrak": 1, "Tidak Tetap": 0}
)

df_encoded["Status_Pernikahan"] = df_encoded["Status_Pernikahan"].map(
    {"Menikah": 2, "Belum Menikah": 1, "Belum": 0}
)

df_encoded["Riwayat_Kredit"] = df_encoded["Riwayat_Kredit"].map(
    {"Baik": 2, "Cukup": 1, "Buruk": 0}
)

df_encoded["Kelayakan_Kredit"] = df_encoded["Kelayakan_Kredit"].map(
    {"Layak": 1, "Tidak": 0}
)

df_encoded
```

```python
x = df_encoded[
    [
        "Usia",
        "Pendapatan_Bulanan",
        "Status_Pekerjaan",
        "Lama_Bekerja",
        "Status_Pernikahan",
        "Jumlah_Tanggungan",
        "Skor_Kredit",
        "Riwayat_Kredit",
    ]
]
y = df_encoded["Kelayakan_Kredit"]
```

```python
model = DecisionTreeClassifier()
model.fit(x, y)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.3, random_state=42
)

y_pred = model.predict(x_test)
cleaned_accuracy = accuracy_score(y_test, y_pred)
print("Akurasi setelah pembersihan data:", cleaned_accuracy)
```
