```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

data = {
    "Name ": [
        "Alice",
        "Bob",
        "Charlie",
        "David",
        "Eve",
        "Frank",
        "Grace",
        "Henry",
        "Ivy",
        "Ikmal",
    ],
    "Value": [88, 43, 95, 67, 78, 82, 90, 55, 80, 92],
    "Class": ["A", "B", "A", "B", "A", "B", "A", "B", "A", "B"],
}

df = pd.DataFrame(data)
df
```

Menampilkan data 5 teratas

```python
df.head(5)
```

Menampilkan data statistik

```python
df.describe(include="all")
```

Menampilkan data kolom nilai

```python
df["Value"]
```

Menampilkan data dari tertinggi hingga terendah

```python
df.sort_values("Value", ascending=False)
```

Menampilkan data dalam bentuk grafik

```python
df.plot(kind="bar", x="Name ", y="Value", title="Value by Name")
plt.show()
```

Menampilkan data atau memfilter mana yang lulus dan tidak

```python
df["Status"] = np.where(df["Value"] >= 80, "Pass", "Fail")
df[df["Class"] == "A"]
```

Menampilkan data rata-rata kelas per tiap kelas (grup)

```python
average_value_of_a_class = df.groupby("Class")["Value"].mean()
average_value_of_a_class.plot(kind="bar", title="Average Value by Class")
```

**Studi Kasus**

Membuat data baru dan menampilkannya

```python
new_data = {
    "Products": ["Coffee", "Tea", "Juice", "Milk", "Bread"],
    "Sum_Sales": [150, 120, 100, 80, 130],
    "Price": [18000, 6000, 10000, 20000, 8000],
}

new_df = pd.DataFrame(new_data)
new_df.head()
```

Menampilkan data statistik 

```python
new_df.describe(include="all")
```

Tambahkan kolom total penjualan 'Jumlah_Terjual' kali 'Harga_Satuan'

```python
new_df["Total_Sales"] = new_df["Sum_Sales"] * new_df["Price"]

print("\nData Total Sales:")
new_df
```

Rata-rata total penjualan

```python
average_total_sales = new_df["Total_Sales"].mean()
print("\nAverage Total Sales:", average_total_sales)
```

**Grafik Total Penjualan**

```python
plt.figure(figsize=(8, 5))
plt.bar(new_df["Products"], new_df["Total_Sales"])

plt.title("Grafik Total Penjualan")
plt.xlabel("Produk")
plt.ylabel("Total Penjualan")

plt.show()
```
