```python
import pandas as pd
import matplotlib.pyplot as plt
from datetime import datetime

data_laptop = {
    "Brand": ["ASUS", "HP", "LENOVO", "APPLE", "ACER"],
    "CPU": [
        "Core i5-13500H",
        "Reyzen 5 7530U",
        "Core i7-14700HX",
        "M2-Pro",
        "Core i3-N305",
    ],
    "RAM": ["16GB", "8GB", "32GB", "16GB", "8GB"],
    "Storage": ["512GB SSD", "512GB SSD", "1TB SSD", "512GB SSD", "512GB SSD"],
    "Price": ["11JT", "8JT", "25JT", "21JT", "5.5JT"],
}

df_laptop = pd.DataFrame(data_laptop)
df_laptop.head()
```

**ANALISA PERUBAHAN KOLOM RELEVAN**

* Perubahan string ke float
* Penambahan kolom baru
* Penambahan kolom analisis (Kategori Harga)

```python
# Menghapus "JT" dan mengubah tipe data menjadi float
df_laptop["Harga (Juta)"] = df_laptop["Price"].str.replace("JT", "").astype(float)
```

```python
# Penambahan kolom baru
df_laptop["Tanggal Data"] = datetime.now().strftime("%Y-%m-%d")
df_laptop["Diskon (%)"] = 10
df_laptop["Harga Final (Juta)"] = df_laptop["Harga (Juta)"] - (
    df_laptop["Harga (Juta)"] * df_laptop["Diskon (%)"] / 100
)
```

```python
# Penambahan kolom analisis
def tentukan_kategori(harga):
    if harga >= 20:
        return "Premium"
    elif harga >= 10:
        return "Menengah"
    else:
        return "Terjangkau"


df_laptop["Kategori Produk"] = df_laptop["Harga Final (Juta)"].apply(tentukan_kategori)
```
